```python
from pyspark.sql import functions as F
df = spark.read.parquet("Files/raw/ruh_flights/")
print("total:", df.count(), "| distinct:", df.dropDuplicates().count())
d = (df.groupBy("flight_number", "`movement.scheduledTime.utc`", "flight_type")
       .count().filter("count > 1"))

df.join(d.limit(3), ["flight_number", "movement.scheduledTime.utc", "flight_type"]) \
  .select("flight_number", "flight_type", "`movement.scheduledTime.utc`",
          "codeshareStatus", "`airline.name`", "`aircraft.reg`") \
  .show(10, truncate=False)
```
